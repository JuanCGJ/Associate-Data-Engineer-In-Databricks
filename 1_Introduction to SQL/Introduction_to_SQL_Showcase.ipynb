## Chapter 1 — Databases, Tables, and Data Types

**SQL (Structured Query Language)** is the language used to communicate with a database. Before writing a single query, this chapter is about understanding the shape of what I'm querying.


### 1.1 — Tables: rows, columns, records, fields

A database is organized into **tables**, and a table has two dimensions:
- **Rows = records** — a specific data observation (e.g. one patron, one order).
- **Columns = fields** — one piece of information about that record (e.g. a name, a date).

**Relational databases** go one step further: they define **relationships between tables**, so instead of repeating a customer's full details on every order row, an `orders` table can reference a `customers` table by a shared key. That's the core advantage of the relational model — less duplication, more consistency.

**Unique identifiers (keys)** are what make a record uniquely findable and are what tables use to relate to one another (e.g. a `customer_id` in `orders` pointing back to `id` in `customers`).


### 1.2 — Naming conventions

The course is opinionated about naming, and I want to actually follow this in my own work:

**Table names should be:**
- Clear and descriptive of the data they contain
- **Plural** (`patrons`, not `patron`)
- Lowercase, with underscores instead of spaces (`employee_hire_years`)

**Field names should be:**
- Lowercase, underscores instead of spaces
- **Singular** (`name`, not `names`)
- Different from the table name itself (avoid `patrons.patron_name`; prefer `patrons.name`)


### 1.3 — Data types

| SQL type | Stores | Popular/flexible variant |
|---|---|---|
| String | A sequence of characters (letters, punctuation) | `VARCHAR` |
| Integer | Whole numbers | `INT` |
| Float | Numbers with a fractional part | `NUMERIC` |

Different types are stored differently and take up different amounts of space, and some operations (e.g. arithmetic aggregates like `AVG`) only make sense on certain types — which is why picking the right type when a table is created matters, not just for storage but for what I can later do with the column.

A **schema** is the blueprint that ties this all together: which tables exist, which columns each one has, what type each column is, and how tables relate to each other.


## Chapter 2 — Writing and Comparing Queries

This chapter moves from *describing* a database to actually *querying* one — SQL is best suited for exactly this: pulling answers out of large datasets that would be painful to scan by hand.


### 2.1 — The first query: `SELECT` / `FROM`

`SELECT` and `FROM` are the two keywords every query starts from. The output of a query is often called the **result set**.

```sql
SELECT name
FROM patrons;
```

**Selecting multiple fields** just means listing them, in whichever order I want them returned:

```sql
SELECT name, card_num, total_fine
FROM patrons;
```

**Selecting everything** with `SELECT *` is convenient for exploring an unfamiliar table, though in real work I'd rather name the columns I actually need — it's more explicit and doesn't break silently if the table gains a new column later.

```sql
SELECT *
FROM patrons;
```


### 2.2 — Aliasing and `DISTINCT`

**Aliasing** with `AS` renames a column in the output, without touching the underlying table — useful for making result sets more readable, especially before handing them off to a report or a BI tool:

```sql
SELECT name AS first_name, year_hired
FROM employees;
```

**`DISTINCT`** removes duplicate rows from the result set. It's easy to get this wrong when several columns are involved, so I want to remember the distinction clearly:

```sql
-- Without DISTINCT: repeats every value that appears more than once
SELECT year_hired
FROM employees;

-- With DISTINCT: each individual year only appears once
SELECT DISTINCT year_hired
FROM employees;

-- DISTINCT on multiple columns: uniqueness is evaluated
-- on the COMBINATION of dept_id + year_hired, not each column alone
SELECT DISTINCT dept_id, year_hired
FROM employees;
```


### 2.3 — Views

A **view** is a virtual table — the saved result of a `SELECT` statement — that automatically reflects updates to the underlying data every time it's queried, instead of freezing a snapshot:

```sql
CREATE VIEW employee_hire_years AS
SELECT id, name, year_hired
FROM employees;

-- Later, query the view exactly like a normal table
SELECT id, name
FROM employee_hire_years;
```

This is a pattern I want to reuse for any query I find myself re-running often — turning a one-off `SELECT` into a named view instead of copy-pasting the same SQL repeatedly.


### 2.4 — SQL flavors: PostgreSQL vs SQL Server

All SQL flavors are used with relational databases, follow the same universal standards, and share the vast majority of their keywords — flavors only differ in the extras layered *on top* of that shared standard, similar to dialects of the same language.

| | PostgreSQL | SQL Server |
|---|---|---|
| License | Free and open-source | Free and paid versions |
| Origin | UC Berkeley | Microsoft |
| Flavor name | "PostgreSQL" (system and flavor share a name) | **T-SQL** |

**Same intent, different syntax** — limiting the number of rows returned:

```sql
-- PostgreSQL
SELECT id, name
FROM employees
LIMIT 2;

-- SQL Server
SELECT TOP(2) id, name
FROM employees;
```

Course's own framing, which I like: *choosing a flavor is a bit like choosing an ice cream flavor — almost any choice is a good one*, since the fundamentals transfer either way.


### 2.5 — Putting it together

The wrap-up example from the course captures most of Chapter 2 in a single query — selecting distinct values, aliasing the result, and limiting the output:

```sql
SELECT DISTINCT genre AS unique_genre
FROM books
LIMIT 15;
```
