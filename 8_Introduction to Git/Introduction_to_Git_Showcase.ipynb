## Chapter 1 — Introduction to Git

### 1.1 — What version control is

Version control is the set of processes and tools used to manage changes to files, programs and directories. It is useful for anything that **changes over time** or **needs to be shared** — code, but also data, reports and configuration.

What it gives me:
- Track files in different states.
- Combine different versions of a file.
- Identify one specific version.
- Revert changes.

**Git** is the most popular version control system for software and data projects: it is open source and scalable. Its key benefits are that it stores everything (nothing is lost), lets me compare files at different points in time, shows what changed, who changed it and when, and lets me go back to a previous version.

### 1.2 — Working in the shell

Git commands run in the **shell** (the terminal): a program for executing commands that also makes it easy to inspect files and directories (a directory is just a folder). Three navigation commands cover most of what Git work needs:

| Command | What it does |
|---|---|
| `pwd` | print the current working directory |
| `ls` | list files and sub-directories |
| `cd <dir>` | change into a directory |

### 1.3 — Creating repos

A **Git repo** is a directory with files and sub-directories **plus Git storage** — a hidden `.git` folder where Git keeps every version. That folder is managed by Git and should never be edited by hand.

Repos let me systematically track versions, revert to previous ones, compare versions at different points in time and collaborate with colleagues.

**Two ways to create one:**

```bash
# 1. Create a brand-new repo (creates the folder too)
git init mental-health-workspace
cd mental-health-workspace
git status

# 2. Convert an existing project into a repo (run inside the folder)
git init
```

After converting an existing project, `git status` lists the files as **untracked**: Git sees them but isn't versioning them until they're added.

**Avoid nested repos:** creating a repo inside another repo leaves two `.git` directories, and it becomes ambiguous which one should record each change.

Setting `main` as the default keeps new repos consistent with GitHub's convention.

### 1.4 — Staging and committing

The Git workflow has three steps:
1. **Edit and save** files on the computer.
2. **Add** them to the **staging area**, which tracks what has been modified and is ready to be saved.
3. **Commit**: Git takes a snapshot of the staged files at that point in time, which I can later compare against or revert to.

The staging area works like a draft: I choose exactly which changes go into the next snapshot, so unrelated edits don't end up in the same commit.

```bash
git add README.md          # stage a single file
git add .                  # stage all modified files in this directory and sub-directories
git commit -m "Adding a README."   # -m: write the log message inline, no text editor
```

A log message should be **short and concise** — it is what I (or a colleague) will read later to understand why the change was made.

## Chapter 2 — Version History

### 2.1 — How Git stores a commit

Every commit has three parts:

| Part | What it holds |
|---|---|
| **Commit** | metadata: author, log message, commit time |
| **Tree** | the names and locations of files and directories — like a dictionary mapping keys to files |
| **Blob** | *Binary Large OBject*: a compressed snapshot of a file's contents, of any kind of data |

Each commit is identified by a **hash** — a 40-character pseudo-random string such as `b22eb75a82a68b9c0f1c45b9f5a9b7abe281683a`. Identical files produce identical hashes, so Git only needs to compare hashes to know whether something changed. That is also what makes sharing data between repos efficient.

### 2.2 — Viewing the history with `git log`

`git log` lists commits from newest to oldest, with hash, author, date and log message. In the pager, **space** shows more commits and **q** returns to the terminal.

### 2.3 — History tips and tricks

As a project grows, `git log` gets long. These options narrow it down to what I actually need:

```bash
git log -3                                  # only the 3 most recent commits
git log report.md                           # history of one file
git log -2 mental_health_survey.csv         # combine both
git log --since='Apr 2 2024'                # commits since a date
git log --since='Apr 2 2024' --until='Apr 11 2024'   # commits in a date range
git show c27fa856                           # details and changes of one commit (8-10 hash chars are enough)
```

`--since` / `--until` accept natural language (`"yesterday"`, `"2 weeks ago"`, `"3 months ago"`) or dates. **ISO format `YYYY-MM-DD` is the safe choice**, because a date like `12-06-2024` can mean 6 December or 12 June depending on system settings, and a format like `"15 Jul, 2024"` is invalid.

### 2.4 — Comparing versions with `git diff`

`git diff` shows the difference between two versions. In the output, `a/` is the older version, `b/` is the newer one, and lines starting with `+` were added (`-` would be removed).

| Command | Compares |
|---|---|
| `git diff` | all unstaged files vs the latest commit |
| `git diff report.md` | one unstaged file vs the latest commit |
| `git diff --staged` | all staged files vs the latest commit |
| `git diff --staged report.md` | one staged file vs the latest commit |
| `git diff 35f4b4d 186398f` | two commits by hash (most recent hash second) |
| `git diff HEAD~1 HEAD` | two commits using `HEAD` syntax |

`HEAD` is the state of the latest commit, `HEAD~1` the one before it, `HEAD~2` the one before that, and so on — no need to look up hashes.

**Exercise — comparing a staged file.** Two new participant rows were added and staged. Before committing, I checked exactly what would go in (from inside `data/`):

### 2.5 — Restoring and reverting

Mistakes happen; Git offers a precise tool for each kind:

| Command | Result |
|---|---|
| `git revert HEAD` | undo all files changed in a commit by creating a new commit |
| `git revert --no-edit HEAD` | same, without opening the text editor |
| `git revert -n HEAD` | revert into the staging area without committing |
| `git checkout HEAD~1 -- report.md` | bring back a single file from a previous commit |
| `git restore --staged report.md` | remove one file from the staging area |
| `git restore --staged` | remove all files from the staging area |

Two points I want to keep in mind:
- `git revert` works on **commits, not individual files**, and it never erases history: it adds a new commit that undoes the old one, so the mistake and its fix both stay traceable. To undo a single file, `git checkout <commit> -- <file>` is the right tool, followed by a normal commit.
- If the text editor (nano) does open, save with `Ctrl + O` then `Enter`, and exit with `Ctrl + X`.

```bash
# Revert one file to its previous version, then save that as a new commit
git checkout HEAD~1 -- report.md
git status
git commit -m "Checkout previous version of report.md"
```

## 🧠 Key Takeaways

- **Version control is for more than code.** Data files, reports and configs change over time and are shared, so they benefit from the same history and rollback.
- **The staging area is a deliberate step.** It lets me build focused commits and catch mistakes (`git diff --staged`) before they're saved.
- **Small commits with clear messages** make `git log` a readable record of why each change happened.
- **Filtered history is faster.** `-n`, a file name, and `--since` / `--until` (with ISO dates) take me straight to the relevant commits.
- **`HEAD~n` syntax** avoids copying hashes when comparing or undoing recent commits.
- **Undo without rewriting history.** `git revert` adds a new commit instead of deleting the old one; `git restore --staged` and `git checkout <commit> -- <file>` handle the single-file cases.
- **Never nest repos, never edit `.git` by hand.**

## 💼 Application in a Data Analyst / Data Engineer Role

- **Pipeline code under version control:** SQL models, PySpark jobs and Databricks notebooks (via Git folders/Repos) all live in Git, so every change to a transformation is reviewable and reversible.
- **Data-quality incidents:** `git log -- <file>` and `git diff HEAD~1 HEAD` quickly answer "what changed in this script or seed file, and when?" when a metric suddenly shifts.
- **Safe rollbacks:** `git revert` restores a broken pipeline to its last good state while keeping an audit trail — important where changes must be traceable.
- **Collaboration:** the same workflow scales to branches, remotes and pull requests, which is how this portfolio repo itself is maintained.
