## Chapter 1 — Python Basics

### 1.1 — Hello Python: shell vs script

Python is a general-purpose, open-source language with a huge package ecosystem, which is why it shows up in everything from web apps to data pipelines. There are two ways to run it:

- **Interactive shell (IPython):** type a command, get the result immediately. Good for exploring.
- **Script (`.py` file):** a list of commands run top to bottom. Good for anything that must be repeatable.

One difference matters in practice: the shell echoes the value of the last expression automatically, but a script only shows what you explicitly `print()`.

````python
# In a script, nothing is shown unless it is printed
print("Hello Python!")
print(7 + 10)
````

### 1.2 — Variables

A variable gives a value a specific, **case-sensitive** name so it can be reused later. The course example: computing the Body Mass Index (BMI = weight / height²) for a person 1.79 m tall weighing 68.7 kg.

````python
height = 1.79   # meters
weight = 68.7   # kilograms

bmi = weight / height ** 2   # ** is the exponent operator
print(bmi)
````

**Interpretation:** a BMI of about 21.44. Because the result is stored in `bmi`, any later step can use it without recomputing.

**Why variables matter — reproducibility:** if the input changes, I only edit one line and re-run. The logic stays the same.

````python
weight = 74.2   # new measurement; everything downstream updates on re-run
bmi = weight / height ** 2
print(bmi)
````

### 1.3 — Python data types

Every value has a type, and the type decides what operations make sense. The four basic ones:

| Type | Meaning | Example |
|---|---|---|
| `float` | real number | `1.79` |
| `int` | integer | `5` |
| `str` | text (single or double quotes) | `"body mass index"` |
| `bool` | logical value | `True` / `False` |

````python
day_of_week = 5
x = "body mass index"
y = 'this works too'
z = True

for value in [bmi, day_of_week, x, y, z]:
    print(repr(value), "->", type(value).__name__)
````

**Different type, different behavior:** the same `+` operator adds numbers but concatenates strings.

````python
print(2 + 3)          # int + int  -> arithmetic sum
print('ab' + 'cd')    # str + str  -> concatenation
````

Mixing types without converting them is a common error, so explicit conversion with `str()`, `int()`, `float()` and `bool()` is part of the toolkit:

````python
savings = 100
growth_multiplier = 1.1
result = savings * growth_multiplier ** 7

# print("I started with $" + savings)  -> TypeError: can only concatenate str (not "int") to str
print("I started with $" + str(savings) + " and now have $" + str(round(result, 2)))

pi_string = "3.1415926"
pi_float = float(pi_string)   # text -> number, so it can be used in calculations
print(pi_float * 2)
````

## Chapter 2 — Python Lists

### 2.1 — Creating lists

One variable per data point doesn't scale: four family members means `height1` … `height4`. A **list** groups a collection of values under one name. Lists can hold any type, mix types, and even contain other lists.

````python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
print(fam)

# A list of lists: one [name, height] pair per person
fam2 = [["liz", 1.73],
        ["emma", 1.68],
        ["mom", 1.71],
        ["dad", 1.89]]
print(fam2)
print(type(fam), type(fam2))
````

**Interpretation:** `fam` is flat, `fam2` is nested — both are of type `list`. The nested form is closer to a table: each inner list is a row.

### 2.2 — Subsetting lists

Python uses **zero-based indexing**: the first element is `[0]`. Negative indices count from the end, so `[-1]` is always the last element.

Slicing uses `[start:end]`, where `start` is **inclusive** and `end` is **exclusive**. Leaving either side empty means "from the beginning" or "to the end".

````python
print(fam[3])     # 4th element  -> 1.68
print(fam[6])     # 'dad'
print(fam[-1])    # last element -> same as fam[7]

print(fam[3:5])   # elements 3 and 4      -> [1.68, 'mom']
print(fam[1:4])   # elements 1, 2 and 3
print(fam[:4])    # first four elements
print(fam[5:])    # from index 5 to the end
````

````python
# Subsetting a list of lists: first pick the row, then the column
print(fam2[2])       # ['mom', 1.71]
print(fam2[2][0])    # 'mom'
print(fam2[-1][1])   # dad's height
````

### 2.3 — Manipulating lists

Three operations: **change** elements (assign to an index or slice), **add** elements (`+`), and **remove** elements (`del`).

````python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]

fam[7] = 1.86                  # change one element
fam[0:2] = ["lisa", 1.74]      # change a slice
print(fam)

fam_ext = fam + ["me", 1.79]   # add elements (creates a new list)
print(fam_ext)

del fam[2]                     # remove an element; indices after it shift left
print(fam)
````

**Watch out:** `del` shifts every later element one position to the left, so indices computed before the deletion no longer point to the same items.

### 2.4 — Behind the scenes: references vs copies

A list variable doesn't hold the values themselves — it holds a **reference** to where the list lives in memory. Assigning `y = x` copies the reference, not the data, so both names point to the same list.

````python
x = ["a", "b", "c"]
y = x            # same list, two names
y[1] = "z"
print("y:", y)
print("x:", x)   # x changed too!
````

````python
x = ["a", "b", "c"]
y = list(x)      # or x[:]  -> an explicit, independent copy
y[1] = "z"
print("y:", y)
print("x:", x)   # x is untouched
````

**Interpretation:** this is one of the most common silent bugs. When a pipeline step must not alter its input, it should work on a copy (`list(x)` or `x[:]`).

## Chapter 3 — Functions and Packages

### 3.1 — Functions

A function is a reusable piece of code that solves a specific task — I call it instead of writing the logic myself. `type()` and `print()` were already functions. If a task is standard, a function for it most likely exists.

````python
fam = [1.73, 1.68, 1.71, 1.89]

tallest = max(fam)        # largest value in the list
print(tallest)
print(len(fam))           # number of elements

print(round(1.68, 1))     # round to 1 decimal  -> 1.7
print(round(1.68))        # ndigits omitted     -> integer 2
````

**Optional arguments:** `round(number, ndigits=None)` has one required and one optional argument. When `ndigits` is left out it defaults to `None` and the result is an integer. `help()` shows this signature:

````python
help(round)
````

````python
# sorted() returns a new, ordered list; reverse is an optional argument
first = [11.25, 18.0, 20.0]
second = [10.75, 9.50]
full = first + second
print(sorted(full, reverse=True))
````

### 3.2 — Methods

Methods are **functions that belong to an object**. In Python everything is an object, and the methods available depend on the object's type: a string has `replace()`, a list does not.

````python
fam = ["liz", 1.73, "emma", 1.68, "mom", 1.71, "dad", 1.89]
sister = "liz"

# list methods
print(fam.index("mom"))   # position of "mom" -> 4
print(fam.count(1.73))    # how many times 1.73 appears -> 1

# str methods
print(sister.capitalize())        # 'Liz'
print(sister.replace("z", "sa"))  # 'lisa'
print(sister.upper())             # 'LIZ'
print(sister.index("z"))          # strings also have index()
````

````python
# Calling a method that doesn't exist for that type raises an error
try:
    fam.replace("mom", "mommy")
except AttributeError as e:
    print("AttributeError:", e)
````

Some methods **change the object in place**. `append()` and `reverse()` modify the list directly and return `None`:

````python
fam.append("me")
fam.append(1.79)
print(fam)

areas = [11.25, 18.0, 20.0, 10.75, 9.50]
areas.reverse()
print(areas)
````

**Summary of the distinction:** a function is called on its own (`type(fam)`, `len(fam)`); a method is called on an object (`fam.index("dad")`).

### 3.3 — Packages

Putting every function into core Python would make it huge, messy and hard to maintain. Instead, extra code lives in **packages**: directories of Python scripts, where each script is a **module** that defines functions, methods and types. Examples: **NumPy** (numeric arrays), **Matplotlib** (plots), **scikit-learn** (machine learning).

Install once from the terminal, then import in each script:

```bash
pip3 install numpy
```

There are three common ways to import:

````python
import numpy                 # full name every time
print(numpy.array([1, 2, 3]))

import numpy as np           # standard alias, used everywhere in practice
print(np.array([1, 2, 3]))

from numpy import array      # import one function only
print(array([1, 2, 3]))
````

````python
# Another example from the standard library: the math package
import math
r = 0.43
C = 2 * math.pi * r      # circumference
A = math.pi * r ** 2     # area
print("Circumference:", round(C, 4))
print("Area:", round(A, 4))
````

**Readability trade-off:** `from numpy import array` is shorter, but further down a long script it's no longer obvious that `array()` comes from NumPy. `import numpy as np` with `np.array(...)` makes the origin explicit, which is why it's the convention I follow.

## Chapter 4 — NumPy

### 4.1 — Why NumPy

Lists are powerful but don't support math over a whole collection. Trying to compute BMI for five people with lists fails:

````python
height = [1.73, 1.68, 1.71, 1.89, 1.79]
weight = [65.4, 59.2, 63.6, 88.4, 68.7]

try:
    weight / height ** 2
except TypeError as e:
    print("TypeError:", e)
````

**NumPy** (Numeric Python) provides the **NumPy array**: an alternative to the list that performs calculations element-wise over the entire array, and does it fast.

````python
import numpy as np

np_height = np.array(height)
np_weight = np.array(weight)

bmi = np_weight / np_height ** 2   # one expression, applied to every element
print(bmi)
````

**Interpretation:** one BMI per person, computed without a loop. This vectorized style is exactly how Spark and pandas columns are processed later in the track.

### 4.2 — NumPy remarks: one type, different behavior

A NumPy array can contain **only one type**. If values are mixed, NumPy converts them all to a common type (here, strings). And because the type is different, operators behave differently from lists:

````python
print(np.array([1.0, "is", True]))   # everything coerced to str

python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])
print(python_list + python_list)     # list: concatenation
print(numpy_array + numpy_array)     # array: element-wise sum
````

### 4.3 — Subsetting NumPy arrays

Indexing works like lists, plus **boolean indexing**: a comparison returns an array of `True`/`False`, which can be used to select only the matching elements.

````python
print(bmi[1])          # second element

over = bmi > 23        # boolean array
print(over)
print(bmi[over])       # only the BMIs above 23
````

**Interpretation:** only one person has a BMI over 23. Boolean masks are the NumPy version of a SQL `WHERE` clause, and the same idea appears as `filter()` in PySpark.

### 4.4 — 2D NumPy arrays

`np.array` creates objects of type `numpy.ndarray`, which can have more than one dimension. A 2D array is like a table: here, row 0 holds heights and row 1 holds weights. `.shape` returns `(rows, columns)`.

````python
print(type(np_height))

np_2d = np.array([[1.73, 1.68, 1.71, 1.89, 1.79],
                  [65.4, 59.2, 63.6, 88.4, 68.7]])
print(np_2d)
print(np_2d.shape)   # 2 rows, 5 columns
````

````python
# A single string turns the whole 2D array into strings
print(np.array([[1.73, 1.68, 1.71, 1.89, 1.79],
                [65.4, 59.2, 63.6, 88.4, "68.7"]]))
````

**Subsetting in 2D** uses `[row, column]`. A colon `:` means "all of them".

````python
print(np_2d[0])          # entire first row (heights)
print(np_2d[0][2])       # row 0, column 2 -> 1.71
print(np_2d[0, 2])       # same thing, preferred syntax
print(np_2d[:, 1:3])     # all rows, columns 1 and 2
print(np_2d[1, :])       # entire second row (weights)
````

````python
# Element-wise operations also work between 2D arrays:
# convert heights from m to cm and weights from kg to lb in one step
conversion = np.array([[100], [2.20462]])
print(np.round(np_2d * conversion, 1))
````

### 4.5 — NumPy: basic statistics

With a handful of values I can just look at the data. With thousands, I need summary statistics. The course scenario: a **city-wide survey** with height and weight for 5,000 people.

`np.random.normal(mean, std, n)` generates normally distributed sample data, and `np.column_stack()` combines the two columns into one 2D array (one row per person).

````python
np.random.seed(42)   # fixed seed so the numbers are reproducible

height = np.round(np.random.normal(1.75, 0.20, 5000), 2)
weight = np.round(np.random.normal(60.32, 15, 5000), 2)
np_city = np.column_stack((height, weight))

print(np_city.shape)
print(np_city[:5])
````

````python
heights = np_city[:, 0]
weights = np_city[:, 1]

print("Mean height:  ", round(np.mean(heights), 4))
print("Median height:", np.median(heights))
print("Std height:   ", round(np.std(heights), 4))
print("Correlation height/weight:")
print(np.round(np.corrcoef(heights, weights), 4))
````

**Interpretation:**
- Mean and median are both close to 1.75 m, which is expected for a symmetric normal distribution. A large gap between them would be a sign of skew or outliers.
- The standard deviation is close to 0.20, the value used to generate the data.
- The correlation is close to 0, because height and weight were generated independently. In real survey data I would expect a positive correlation, so a near-zero value would be worth investigating.

Other built-ins such as `np.sum()` and `np.sort()` work the same way. They are fast because the array enforces a **single data type**.

````python
# Sanity check typical of data cleaning: physically impossible values
print("Heights <= 0:", np.sum(heights <= 0))
print("Min / max height:", heights.min(), "/", heights.max())
print("Median weight of people taller than 2 m:", np.median(weights[heights > 2]))
````

## 🧠 Key Takeaways

- **Types drive behavior.** The same operator (`+`) concatenates lists and strings but adds numbers and arrays. Knowing the type of each value prevents silent errors.
- **Lists are references.** `y = x` shares the same list; `list(x)` or `x[:]` makes an independent copy.
- **Reuse before you write.** For standard tasks, a function, method or package already exists. `help()` documents its arguments.
- **`import numpy as np`** is the readable convention because it makes the origin of each function explicit.
- **Vectorize instead of looping.** NumPy applies one expression to every element, and boolean masks filter data the same way a `WHERE` clause does.
- **Summarize large data.** Mean, median, standard deviation and correlation are the first look at a dataset, and comparing them already reveals skew, outliers or unexpected relationships.
